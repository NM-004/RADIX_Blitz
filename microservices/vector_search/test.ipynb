# Vector Search API Test

This notebook tests the `/vector-match` endpoint of the vector search microservice.

Make sure you have started the FastAPI server in the terminal first:
```bash
cd microservices/vector_search
uvicorn main:app --reload --port 8001
```

````python
import requests
import json

url = "http://127.0.0.1:8001/vector-match"

# Sample candidate and company requirements
payload = {
    "candidate_id": "cand_123",
    "candidate_skills": ["Python", "React", "Machine Learning", "Postgres Database"],
    "company_id": "comp_456",
    "required_skills": ["Python Programming", "React JS", "SQL", "Docker"]
}

try:
    response = requests.post(url, json=payload)
    print(f"Status Code: {response.status_code}\n")
    print(json.dumps(response.json(), indent=2))
except requests.exceptions.ConnectionError:
    print("Error: Connection refused. Is the FastAPI server running on port 8001?")
````
